```python
print("\n Defining utility functions...")

def save_detection_json(filename, detections, output_path):
    """Save detection results to JSON file"""
    result = {
        "filename": filename,
        "detections": detections
    }

    json_path = output_path / f"{Path(filename).stem}.json"
    with open(json_path, 'w') as f:
        json.dump(result, f, indent=2)

    return json_path

def load_test_images(folder_path, limit=5):
    """Load sample images for testing"""
    image_extensions = ['*.jpg', '*.jpeg', '*.png']
    images = []

    for ext in image_extensions:
        images.extend(glob.glob(str(folder_path / ext)))

    return images[:limit]

print(" Utility functions defined")
```

```python
print("\n  DATASET VISUALIZATION")
print("-" * 30)

def show_sample_images(dataset_dir, split='train', num_samples=3):
    """Display sample images from dataset"""
    images_dir = dataset_dir / split / 'images'
    image_files = list(images_dir.glob('*.jpg'))[:num_samples]

    if len(image_files) == 0:
        print(f"No images found in {images_dir}")
        return

    fig, axes = plt.subplots(1, num_samples, figsize=(15, 5))
    if num_samples == 1:
        axes = [axes]

    for i, img_path in enumerate(image_files):
        # Load and display image
        img = cv2.imread(str(img_path))
        if img is not None:
            img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            axes[i].imshow(img_rgb)
            axes[i].set_title(f'{split.capitalize()} Sample {i+1}')
            axes[i].axis('off')
        else:
            axes[i].text(0.5, 0.5, 'Image not found', ha='center', va='center')
            axes[i].axis('off')

    plt.tight_layout()
    plt.show()
```

```python
print("\n STARTING YOLOv8x TRAINING FOR MAXIMUM ACCURACY")
print("=" * 55)

# Record start time
start_time = time.time()
start_datetime = datetime.now()

print(f" Training started at: {start_datetime.strftime('%Y-%m-%d %H:%M:%S')}")
print(f"  Estimated duration: 2-3 hours")
print(f" Target accuracy: >95%")

print("\n" + "="*55)
print("TRAINING IN PROGRESS - LIVE METRICS BELOW")
print("="*55)

try:
    # Start YOLOv8x training
    results = model.train(
        data=TRAINING_CONFIG['data'],
        epochs=TRAINING_CONFIG['epochs'],
        imgsz=TRAINING_CONFIG['imgsz'],
        batch=TRAINING_CONFIG['batch'],
        device=TRAINING_CONFIG['device'],
        project=TRAINING_CONFIG['project'],
        name=TRAINING_CONFIG['name'],
        save=TRAINING_CONFIG['save'],
        plots=TRAINING_CONFIG['plots'],
        verbose=TRAINING_CONFIG['verbose'],
        patience=TRAINING_CONFIG['patience'],
        save_period=TRAINING_CONFIG['save_period'],
        workers=TRAINING_CONFIG['workers'],
        optimizer=TRAINING_CONFIG['optimizer'],
        lr0=TRAINING_CONFIG['lr0'],
        weight_decay=TRAINING_CONFIG['weight_decay'],
        warmup_epochs=TRAINING_CONFIG['warmup_epochs'],
        close_mosaic=TRAINING_CONFIG['close_mosaic'],
        mixup=TRAINING_CONFIG['mixup'],
        copy_paste=TRAINING_CONFIG['copy_paste']
    )

    # Calculate training duration
    end_time = time.time()
    end_datetime = datetime.now()
    training_duration = (end_time - start_time) / 3600  # Convert to hours

    print("\n" + "=" * 55)
    print(" YOLOv8x TRAINING COMPLETED SUCCESSFULLY!")
    print("=" * 55)
    print(f" Completed at: {end_datetime.strftime('%Y-%m-%d %H:%M:%S')}")
    print(f"  Total training time: {training_duration:.2f} hours")
    print(f" Results saved to: {RESULTS_DIR}/glove_detection_yolov8x/")

    # Show best model path
    best_model_path = RESULTS_DIR / "glove_detection_yolov8x/weights/best.pt"
    print(f" Best model saved to: {best_model_path}")

    # Display final metrics
    print(f"\n FINAL TRAINING RESULTS:")
    print("-" * 30)
    if hasattr(results, 'results_dict'):
        metrics = results.results_dict
        print(f"   mAP50: {metrics.get('metrics/mAP50(B)', 'N/A')}")
        print(f"   mAP50-95: {metrics.get('metrics/mAP50-95(B)', 'N/A')}")
        print(f"   Precision: {metrics.get('metrics/precision(B)', 'N/A')}")
        print(f"   Recall: {metrics.get('metrics/recall(B)', 'N/A')}")

    print(f"\n MAXIMUM ACCURACY ACHIEVED!")
    print(f" YOLOv8x training phase complete!")
    print(f" Next: Build detection script for submission")

except Exception as e:
    print(f"\n TRAINING ERROR: {str(e)}")
    print(f"\n Troubleshooting suggestions:")
    print(f"   1. Reduce batch size: batch=8 or batch=4")
    print(f"   2. Reduce epochs: epochs=20")
    print(f"   3. Check GPU memory: !nvidia-smi")
    print(f"   4. Restart runtime and try again")

    # Show partial results if available
    if 'results' in locals():
        print(f"\n Partial results available")

print("\n" + "=" * 55)
```

```python
print(" BUILDING DETECTION SCRIPT WITH  MY IMPROVED MODEL")
print("=" * 50)

# Load My BEST trained model (15 epochs)
best_model_path = Path("/content/training_results/glove_detection_yolov8s_improved/weights/best.pt")
trained_model = YOLO(str(best_model_path))

print(f" Loaded improved model: {best_model_path}")
print(f" Performance: 59.6% mAP50, 73.3% Precision")

# Create output directories for assessment
OUTPUT_DIR = Path("/content/output")
LOGS_DIR = Path("/content/logs")
OUTPUT_DIR.mkdir(exist_ok=True)
LOGS_DIR.mkdir(exist_ok=True)

print(f" Output directories created:")
print(f"   Images: {OUTPUT_DIR}")
print(f"   Logs: {LOGS_DIR}")

# Detection functions
def save_detection_json(filename, detections, output_path):
    """Save detection results to JSON file as required by assessment"""
    result = {
        "filename": filename,
        "detections": detections
    }

    json_path = output_path / f"{Path(filename).stem}.json"
    with open(json_path, 'w') as f:
        json.dump(result, f, indent=2)

    return json_path

def process_detections(results, confidence_threshold=0.5):
    """Convert YOLO results to assessment format"""
    detections = []

    for result in results:
        boxes = result.boxes
        if boxes is not None:
            for i in range(len(boxes)):
                conf = float(boxes.conf[i])
                cls = int(boxes.cls[i])
                xyxy = boxes.xyxy[i].tolist()

                if conf >= confidence_threshold:
                    # Map class to assessment labels
                    label = "Wearing_Gloves" if cls == 1 else "No_Gloves"

                    detection = {
                        "label": label,
                        "confidence": round(conf, 2),
                        "bbox": [round(x, 1) for x in xyxy]  # [x1, y1, x2, y2]
                    }
                    detections.append(detection)

    return detections

print(" Detection functions ready!")
print("\n Ready to test detection on sample images...")
```

```python
#!/usr/bin/env python3
"""
Complete YOLOv8m Training Script
Run this file to train your glove detection model
"""

import time
from datetime import datetime
from pathlib import Path

# Step 1: Install required packages (run in terminal first)
"""
pip install ultralytics
pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu118
"""

# Step 2: Import required libraries
try:
    from ultralytics import YOLO
    import torch
    print("✅ All libraries imported successfully")
    print(f"🔥 CUDA available: {torch.cuda.is_available()}")
    if torch.cuda.is_available():
        print(f"🎮 GPU: {torch.cuda.get_device_name(0)}")
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("🔧 Run: pip install ultralytics torch")
    exit(1)

# Step 3: Setup paths (UPDATE THESE FOR YOUR PROJECT)
DATASET_PATH = "path/to/your/dataset.yaml"  # ⚠️ UPDATE THIS PATH
PROJECT_DIR = "runs/detect"
MODEL_NAME = "glove_detection_yolov8m"

# Step 4: Create dataset.yaml example (if you don't have one)
def create_sample_dataset_yaml():
    """Creates a sample dataset.yaml file - modify for your data"""
    yaml_content = """# YOLOv8 Dataset Configuration
path: /path/to/your/dataset  # dataset root dir (UPDATE THIS)
train: images/train  # train images (relative to 'path')
val: images/val      # val images (relative to 'path')
test: images/test    # test images (optional)

# Classes
nc: 1  # number of classes (UPDATE: 1 for gloves, or more if multiple classes)
names: ['glove']  # class names (UPDATE THESE)

# Example for multiple classes:
# nc: 3
# names: ['glove', 'hand', 'tool']
"""

    with open('dataset.yaml', 'w') as f:
        f.write(yaml_content)
    print("📝 Created sample dataset.yaml - please update the paths!")
    return 'dataset.yaml'

# Step 5: Main training function
def train_yolov8m():
    """Main training function"""

    # Check if dataset.yaml exists
    if not Path(DATASET_PATH).exists():
        print(f"⚠️  Dataset file not found: {DATASET_PATH}")
        print("🔧 Creating sample dataset.yaml...")
        dataset_path = create_sample_dataset_yaml()
        print(f"✏️  Please edit {dataset_path} with your actual paths before running!")
        return False

    # Start training
    print("\n🚀 STARTING YOLOv8m TRAINING (30 EPOCHS)")
    print("=" * 60)

    start_time = time.time()
    start_datetime = datetime.now()
    print(f"⏰ Started: {start_datetime.strftime('%Y-%m-%d %H:%M:%S')}")

    try:
        # Initialize YOLOv8m model
        print("📥 Loading YOLOv8m model...")
        model = YOLO('yolov8m.pt')  # Will download automatically if not present

        # Training configuration
        TRAINING_CONFIG = {
            'data': DATASET_PATH,  # Your dataset.yaml path
            'epochs': 30,
            'imgsz': 640,
            'batch': 16,  # Adjust based on GPU memory
            'workers': 4,
            'device': 0,  # 0 for GPU, 'cpu' for CPU
            'project': PROJECT_DIR,
            'name': MODEL_NAME,
            'exist_ok': True,
            'pretrained': True,
            'optimizer': 'SGD',
            'verbose': True,
            'seed': 42,
            'deterministic': True,
            'single_cls': False,
            'rect': False,
            'cos_lr': False,
            'close_mosaic': 10,
            'resume': False,
            'amp': True,  # Automatic Mixed Precision
            'fraction': 1.0,
            'profile': False,
            'freeze': None,
            'lr0': 0.01,
            'lrf': 0.01,
            'momentum': 0.937,
            'weight_decay': 0.0005,
            'warmup_epochs': 3.0,
            'warmup_momentum': 0.8,
            'warmup_bias_lr': 0.1,
            'box': 7.5,
            'cls': 0.5,
            'dfl': 1.5,
            'pose': 12.0,
            'kobj': 1.0,
            'label_smoothing': 0.0,
            'nbs': 64,
            'hsv_h': 0.015,
            'hsv_s': 0.7,
            'hsv_v': 0.4,
            'degrees': 0.0,
            'translate': 0.1,
            'scale': 0.5,
            'shear': 0.0,
            'perspective': 0.0,
            'flipud': 0.0,
            'fliplr': 0.5,
            'mosaic': 1.0,
            'mixup': 0.0,
            'copy_paste': 0.0
        }

        # Start training
        print(f"🔥 Training YOLOv8m with batch size {TRAINING_CONFIG['batch']}")
        print("🎯 This will take several hours for 30 epochs...")

        results = model.train(**TRAINING_CONFIG)

        # Training completed
        end_time = time.time()
        duration = (end_time - start_time) / 3600

        print(f"\n✅ YOLOv8m TRAINING COMPLETED!")
        print(f"⏱️  Duration: {duration:.2f} hours")
        print(f"📁 Results: {PROJECT_DIR}/{MODEL_NAME}/")

        # Show model paths
        results_dir = Path(PROJECT_DIR) / MODEL_NAME
        best_model = results_dir / "weights" / "best.pt"
        last_model = results_dir / "weights" / "last.pt"

        print(f"🏆 Best model: {best_model}")
        print(f"📊 Last model: {last_model}")

        # Show final metrics
        if hasattr(results, 'results_dict'):
            metrics = results.results_dict
            print(f"\n📊 Final Results:")
            print(f"   📈 mAP50: {metrics.get('metrics/mAP50(B)', 'N/A')}")
            print(f"   🎯 Precision: {metrics.get('metrics/precision(B)', 'N/A')}")
            print(f"   🔍 Recall: {metrics.get('metrics/recall(B)', 'N/A')}")

        # Show training curves location
        print(f"\n📈 View training curves: {results_dir}/")
        print("   - results.png (training curves)")
        print("   - confusion_matrix.png")
        print("   - val_batch0_labels.jpg (validation samples)")

        print(f"\n🎉 Training complete! Use {best_model} for inference!")
        return True

    except Exception as e:
        print(f"❌ Training Error: {str(e)}")
        print("\n🔧 Common fixes:")
        print("   • Reduce batch size: change 'batch': 16 to 'batch': 8")
        print("   • Check dataset.yaml paths are correct")
        print("   • Ensure images and labels are in correct format")
        print("   • Verify CUDA/GPU setup if using GPU")
        return False

# Step 6: Quick validation function
def validate_setup():
    """Validate the setup before training"""
    print("🔍 Validating setup...")

    # Check dataset file
    if not Path(DATASET_PATH).exists():
        print(f"❌ Dataset file missing: {DATASET_PATH}")
        return False

    # Check CUDA if using GPU
    if torch.cuda.is_available():
        print(f"✅ GPU available: {torch.cuda.get_device_name(0)}")
        print(f"   CUDA version: {torch.version.cuda}")
        print(f"   GPU memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
    else:
        print("⚠️  No GPU detected - training will use CPU (much slower)")

    return True

# Step 7: Run the training
if __name__ == "__main__":
    print("🤖 YOLOv8m Glove Detection Training")
    print("=" * 60)

    # Update this path to your actual dataset.yaml
    DATASET_PATH = "dataset.yaml"  # ⚠️ UPDATE THIS

    # Validate setup
    if not validate_setup():
        print("\n🔧 Please fix the issues above before training")
        exit(1)

    # Ask user confirmation
    response = input(f"\n🚀 Ready to train for 30 epochs? This will take 3-6 hours. Continue? (y/N): ")
    if response.lower() != 'y':
        print("👋 Training cancelled")
        exit(0)

    # Start training
    success = train_yolov8m()

    if success:
        print("\n🎉 All done! Check the results folder for your trained model.")
    else:
        print("\n❌ Training failed. Check the error messages above.")

# Additional notes
print(f"\n📝 Next steps after training:")
print("1. Use the best.pt model for inference")
print("2. Test with: model = YOLO('runs/detect/glove_detection_yolov8m/weights/best.pt')")
print("3. Run detection: results = model('path/to/image.jpg')")
```

```python
#!/usr/bin/env python3
"""
Fix Dataset Setup for YOLOv8m Training
This will help you create the correct dataset structure and yaml file
"""

import os
from pathlib import Path
import yaml

def create_dataset_structure():
    """Create proper dataset folder structure"""

    print("📁 Creating dataset structure...")

    # Define dataset structure
    dataset_root = Path("glove_dataset")
    folders = [
        "images/train",
        "images/val",
        "images/test",
        "labels/train",
        "labels/val",
        "labels/test"
    ]

    # Create folders
    for folder in folders:
        folder_path = dataset_root / folder
        folder_path.mkdir(parents=True, exist_ok=True)
        print(f"✅ Created: {folder_path}")

    return dataset_root

def create_correct_dataset_yaml(dataset_root):
    """Create properly formatted dataset.yaml"""

    # Get absolute path
    abs_path = dataset_root.absolute()

    # Create dataset configuration
    dataset_config = {
        'path': str(abs_path),  # Absolute path to dataset root
        'train': 'images/train',  # Relative to path above
        'val': 'images/val',      # Relative to path above
        'test': 'images/test',    # Relative to path above (optional)
        'nc': 1,  # Number of classes
        'names': ['glove']  # Class names
    }

    # Save yaml file
    yaml_file = dataset_root / "dataset.yaml"
    with open(yaml_file, 'w') as f:
        yaml.dump(dataset_config, f, default_flow_style=False)

    print(f"✅ Created: {yaml_file}")
    print(f"📝 Dataset path: {abs_path}")

    return yaml_file

def show_example_files():
    """Show examples of what files should look like"""

    print("\n📋 Example label file format (YOLO format):")
    print("labels/train/image001.txt:")
    print("0 0.5 0.5 0.3 0.4")
    print("# Format: class_id center_x center_y width height (all normalized 0-1)")

    print("\n📷 Image and label file naming:")
    print("images/train/image001.jpg  ←→  labels/train/image001.txt")
    print("images/train/image002.png  ←→  labels/train/image002.txt")

def create_sample_files(dataset_root):
    """Create sample placeholder files"""

    print("\n📝 Creating sample placeholder files...")

    # Create sample label files
    sample_label = "0 0.5 0.5 0.3 0.4\n"  # Sample glove detection

    for split in ['train', 'val']:
        label_file = dataset_root / f"labels/{split}/sample.txt"
        with open(label_file, 'w') as f:
            f.write(sample_label)
        print(f"✅ Created sample: {label_file}")

    # Create readme with instructions
    readme_content = """# Glove Detection Dataset

## Structure:
- images/train/ - Training images (.jpg, .png)
- images/val/   - Validation images
- labels/train/ - Training labels (.txt, YOLO format)
- labels/val/   - Validation labels

## Label Format (YOLO):
Each .txt file contains one line per object:
class_id center_x center_y width height

All values normalized to 0-1:
- center_x, center_y: object center relative to image size
- width, height: object size relative to image size
- class_id: 0 for 'glove'

## Example:
For image 'glove001.jpg' with glove at center, half image size:
glove001.txt content: 0 0.5 0.5 0.5 0.5
"""

    readme_file = dataset_root / "README.md"
    with open(readme_file, 'w') as f:
        f.write(readme_content)

    print(f"✅ Created: {readme_file}")

def update_training_script(dataset_yaml_path):
    """Show how to update the training script"""

    print(f"\n🔧 UPDATE YOUR TRAINING SCRIPT:")
    print(f"Change this line in your Python script:")
    print(f'DATASET_PATH = "{dataset_yaml_path.absolute()}"')

    print(f"\nOr run training directly with:")
    print(f"from ultralytics import YOLO")
    print(f"model = YOLO('yolov8m.pt')")
    print(f"results = model.train(data='{dataset_yaml_path.absolute()}', epochs=30)")

def quick_download_sample_dataset():
    """Show commands to download a sample dataset for testing"""

    print(f"\n🚀 QUICK TEST WITH SAMPLE DATASET:")
    print("If you want to test with a sample dataset first:")
    print("")
    print("# Download COCO8 sample dataset")
    print("from ultralytics import YOLO")
    print("model = YOLO('yolov8m.pt')")
    print("results = model.train(data='coco8.yaml', epochs=3)  # Quick test")
    print("")
    print("This will automatically download a small test dataset!")

def main():
    """Main function to set up dataset"""

    print("🔧 YOLOv8 Dataset Setup Helper")
    print("=" * 50)

    # Create dataset structure
    dataset_root = create_dataset_structure()

    # Create correct yaml file
    yaml_file = create_correct_dataset_yaml(dataset_root)

    # Create sample files
    create_sample_files(dataset_root)

    # Show examples
    show_example_files()

    # Show next steps
    update_training_script(yaml_file)

    # Show quick test option
    quick_download_sample_dataset()

    print(f"\n✅ SETUP COMPLETE!")
    print(f"📁 Dataset folder: {dataset_root.absolute()}")
    print(f"📝 Config file: {yaml_file.absolute()}")

    print(f"\n📋 NEXT STEPS:")
    print("1. 📷 Add your images to images/train/ and images/val/")
    print("2. 🏷️  Add corresponding label files to labels/train/ and labels/val/")
    print("3. 🔧 Update your training script with the correct dataset path")
    print("4. 🚀 Run training again!")

    print(f"\n💡 OR test with sample data first:")
    print("   model.train(data='coco8.yaml', epochs=3)")

if __name__ == "__main__":
    main()

# Example of corrected training code snippet
print(f"\n" + "="*50)
print("🔥 CORRECTED TRAINING CODE:")
print("="*50)

corrected_code = '''
from ultralytics import YOLO
import time
from datetime import datetime

# Use the created dataset
DATASET_PATH = "glove_dataset/dataset.yaml"  # Update this path

def train_yolov8m_fixed():
    print("🚀 STARTING YOLOv8m TRAINING (30 EPOCHS)")

    start_time = time.time()
    start_datetime = datetime.now()
    print(f"⏰ Started: {start_datetime.strftime('%Y-%m-%d %H:%M:%S')}")

    try:
        # Load model
        model = YOLO('yolov8m.pt')

        # Start training with corrected dataset path
        results = model.train(
            data=DATASET_PATH,  # Correct path
            epochs=30,
            imgsz=640,
            batch=16,  # Reduce to 8 if memory issues
            device=0,
            project='runs/detect',
            name='glove_detection_yolov8m',
            exist_ok=True
        )

        print("✅ Training completed successfully!")
        return results

    except Exception as e:
        print(f"❌ Error: {e}")
        return None

# Run training
if __name__ == "__main__":
    train_yolov8m_fixed()
'''

print(corrected_code)
```
